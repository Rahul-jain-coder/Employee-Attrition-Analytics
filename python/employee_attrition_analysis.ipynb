# Department Analysis

```python
df.groupby("Department")["MonthlyIncome"].mean()
```

### Observation

- Average monthly income varies across departments.
- Salary differences may be influenced by job responsibilities and organizational hierarchy.

```python
df.groupby("Department")["JobSatisfaction"].mean()
```

### Observation

- Job satisfaction levels differ among departments.
- Departments with lower satisfaction may require management attention.

```python
pd.crosstab(df["Department"],df["Attrition"])
```

### Observation

- Attrition is not equally distributed across departments.
- Some departments experience higher employee turnover than others.

# Attrition Analysis

```python
pd.crosstab(df["OverTime"],df["Attrition"])
```

### Observation

- Employees working overtime tend to have higher attrition compared to employees who do not work overtime.
- Overtime appears to be an important factor influencing employee turnover.

```python
pd.crosstab(df["Gender"],df["Attrition"])
```

### Observation

- Attrition occurs across both genders.
- Any noticeable difference should be further investigated to identify underlying causes.

```python
pd.crosstab(df["MaritalStatus"],df["Attrition"])
```

### Observation

- Attrition varies among different marital status groups.
- Single employees generally experience higher attrition than married employees.

```python
df[df["Attrition"]=="Yes"]
```

### Observation
